1. Demographic & Personal Characteristics

-PER (Numeric): Direct count of people living in the household (e.g., 1, 2, 4).

-ZADULT (Numeric): Direct count of adult residents in the home.

-ELDER (Numeric): Direct count of senior residents aged 65 or older.

-HLTH (Categorical): Primary householder’s self-reported overall health status rating (Excellent, Very Good, Fair, Poor).

-EDUCATION (Categorical, formerly HHGRAD): Educational attainment level of the primary householder (High school graduate, SCHOOLDROPOUT, UNIDROPOUT, Diploma, GRADUATEPROGRAM).

2. Geographic, Location & Climate Indicators

-DEGREE (Categorical): Climate zone classification based on local heating and cooling degree days (Coldest, Cold, Cool, Mild, Mixed, Hot).

-LOCATIONDEVISION (Categorical, formerly DIVISION): U.S. Census geographic region or division where the property is located (New England, Middle Atlantic, East North Central, West North Central, South Atlantic & East South Central, West South Central, Mountain & Pacific).

-CENTALCITYINDICATOR (Categorical, formerly METRO3): Metropolitan Statistical Area (MSA) urban, suburban, or rural classification (Central MSA city, MSA, urban, MSA, rural, Non-MSA, urban, Non-MSA, rural).

3. Income, Costs & Financial Metrics

-ZINC2 (Numeric): Total annual household income expressed directly in dollars (e.g., 45000 = $45,000 annual income).

-ZSMHC (Numeric): Total calculated monthly housing costs in dollars per month (mortgage/rent + utilities + insurance).

-VALUE (Numeric): Estimated market value of the home in dollars for owner-occupied properties (e.g., 250000 = $250,000).

-OWNERSHIP (Categorical, formerly CONDO): Ownership structure type of the housing unit (Condo for condominium/co-op, Not Condo for standard residential property).

-PROPERTYINSURANCE (Categorical, formerly BUYI): Status indicating whether the household maintains homeowners or property insurance (Yes, No).

4. Structure, Physical

-Characteristics & Foundation
BUILT (Numeric/Year): Exact four-digit year of construction (e.g., 1995) or Census multi-year era code.

-HOUSETYPE (Categorical, formerly TYPE): Structural classification or building type of the dwelling (House, apartment, flat, Mobile home with permanent room added, Mobile home, no permanent room added, Housing unit in transient hotel, Housing unit in non transient hotel, Housing unit rooming house, Boat or recreational vehicle, Other).

-BASEMENTQUALITY (Categorical, formerly CELLAR): Foundation or basement design supporting the structure (Basement under all of the house, Basement under part of the house, Crawl space, On a concrete slab, In some other way).

5. Amenities, Equipment & Specific Room Usage

-BUSPER (Numeric): Direct count of rooms inside the housing unit dedicated to business or commercial purposes (e.g., 0 = none, 1 = 1 room).

-DISH (Categorical): Presence of a working dishwasher in the unit (Yes, No).

-COOK (Categorical): Presence of functional primary cooking equipment such as a stove or range (Yes, No).

-GARAGE (Categorical): Presence of an attached or detached garage/carport (Yes, No).

-FIREPLACE (Categorical, formerly FPLWK): Presence of a usable/working fireplace in the home (Yes, No).

# Drop Columns Not needed, And Replace Redundant rows and columns with a Suitable Name:

```python
# Drop DATE:
df = df.drop('DATE', axis=1)
```

## Replace Education Names

```python
# School Dropout:
schoolDropout = ['9th grade','10th grade','12th grade','7th or 8th grade','1st, 2nd, 3rd, or 4th grade','Less than 1st grade','5th or 6th grade','11th grade']
for i in schoolDropout:
  df['HHGRAD'].replace(i,'SCHOOLDROPOUT', inplace=True)
```

```python
# course replaced with degree:
df['HHGRAD'].replace('Professional school degree','Diploma', inplace=True)
```

```python
GradProgram = ['Associate degree in occupational program',"Bachelor's degree","Master's degree",'Associate degree in academic program','Doctorate degree'];
for i in GradProgram:
  df['HHGRAD'].replace(i,'GRADUATEPROGRAM', inplace=True)
```

```python
df['HHGRAD'].replace('Some college but no degree','UNIDROPOUT', inplace=True)
```

## Replace Object Coulumn Names:

```python
# Replace the HHGRAD to EDUCATION:
df['EDUCATION'] = df['HHGRAD']
df = df.drop('HHGRAD', axis=1)
```

```python
# Replace the FPLWK to FIREPLACE:
df['FIREPLACE'] = df['FPLWK']
df = df.drop('FPLWK', axis=1)
```

```python
# Replace the CONDO to OWNERSHIP:
df['OWNERSHIP'] = df['CONDO']
df = df.drop('CONDO', axis=1)
```

```python
# Replac the CELLAR with BASMENTQUALITY
df['BASEMENTQUALITY'] = df['CELLAR']
df = df.drop('CELLAR', axis=1)
```

```python
# Replace the TYPE with HOUSETYPE:
df['HOUSETYPE'] = df['TYPE']
df = df.drop('TYPE', axis=1)
```

```python
# Replace the BUYI with PROPERTYINSURANCE:
df['PROPERTYINSURANCE'] = df['BUYI']
df = df.drop('BUYI', axis=1)
```

```python
# Replace the DEVISION with LOCATIONDEVISION:
df['LOCATIONDEVISION'] = df['DIVISION']
df = df.drop('DIVISION', axis=1)
```

```python
# Replace the METRO3 with CENTALCITYINDICATOR
df['CENTALCITYINDICATOR'] = df['METRO3']
df = df.drop('METRO3', axis=1)
```

## Replace Number Columns:

```python
df = df.rename(columns={
    'PER': 'HOUSEHOLDNUMBER',
    'ZADULT': 'ADULTAMOUNT',
    'ZINC2': 'HOUSEHOLDINCOME',
    'ZSMHC': 'MONTHLYEXPENSES',
    'BUILT': 'BUILTYEAR',
    'BUSPER': 'COMMERCIALROOMS',
    'VALUE': 'PROPERTY',
})
```

# Logistic Regression with RandomForest Classification

```python
from sklearn.preprocessing import LabelEncoder
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from xgboost import XGBClassifier

le = LabelEncoder()
for i in df:
  df[i] = le.fit_transform(df[i])
```

```python
# Logistic Regression

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

x = df.drop('HOUSETYPE', axis=1)
y = df['HOUSETYPE']

x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=42)

model = LogisticRegression()
model.fit(x_train,y_train)

# for the accuracy:
y_pred = model.predict(x_test)
print('Accuracy: ',accuracy_score(y_test,y_pred)*100)
print(classification_report(y_test,y_pred))
sns.heatmap(confusion_matrix(y_test,y_pred), annot=True, cmap='coolwarm')
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import confusion_matrix, accuracy_score, classification_report

rt = RandomForestClassifier(
    n_estimators = 300,
    random_state=42
)
rt.fit(x_train, y_train)
y_pred_rt = rt.predict(x_test)

print('Accuracy_Score' , accuracy_score(y_test, y_pred_rt))
print('\nClassifiction_Report\n')
print(classification_report(y_test, y_pred_rt))

print('\nConfusion_Matrix\n')
sns.heatmap(confusion_matrix(y_test,y_pred_rt) , annot=True , cmap='Blues')
plt.show()
```
